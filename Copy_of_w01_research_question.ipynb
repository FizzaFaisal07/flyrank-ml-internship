## 2. ## Research Question

**Research Question:**
Can machine learning identify which web pages should be prioritized for content refresh based on historical search performance?

**Decision:**
Determine which pages should be refreshed first.

**Unit of Analysis:**
Individual web pages.

**Output:**
A ranked list of pages that are most likely to benefit from a content refresh.

**Action:**
An SEO team reviews and updates the highest-priority pages to improve search performance.

**Cost of a Wrong Recommendation:**
If an important declining page is missed, traffic and visibility may continue to decrease. If a healthy page is selected unnecessarily, time and resources are wasted.

**Why Machine Learning Helps:**
Many factors influence page performance, including search volume, CTR, position, impressions, and content characteristics. Machine learning can combine these signals more effectively than a simple rule-based approach.

```python
import os
import subprocess
import pandas as pd

REPO_URL = "https://github.com/FizzaFaisal07/flyrank-ml-internship.git"
REPO_DIR = "flyrank-ml-internship"

if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", REPO_URL], check=True)

os.chdir(REPO_DIR)

print("Working Directory:", os.getcwd())

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

print("Dataset Overview")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumn Names:")
print(df.columns.tolist())

print("Trend Direction Distribution\n")
print(df["trend_direction"].value_counts())

print("\nUnique Content Types:", df["content_type"].nunique())
```
